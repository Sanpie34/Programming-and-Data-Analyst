# Rock-Paper-Scissors

```python
import numpy as np
```

```python
print("------------------------------")
print("")
print("Welcome to Rock-Paper-Scissors!")
print("Rules: Rock beats Scissors, Scissors beats Paper, Paper beats Rock.")
print("Let's play!")
print("")
print("------------------------------")
print("")
print("Make your choice:")
print("Note: Please capitalize the first letter of your choice (example: Rock, Paper, or Scissors).")
Player = str(input("Choose Rock, Paper, or Scissors:"))
print(f"You chose: {Player}")
Options = ["Rock", "Paper", "Scissors"]
Computer = np.random.choice(Options)
print(f"Computer chose: {Computer}")
print("")
print("------------------------------")
print("")
print("------------- Result -------------")
if Player == Computer:
    print("It's a tie!")
    print("Try again!")
elif (Player == "Rock" and Computer == "Scissors") or (Player == "Paper" and Computer == "Rock") or (Player == "Scissors" and Computer == "Paper"):
    print("You win!")
    print("Congratulations!")
elif Player not in Options:
    print("Invalid choice! Please write Rock, Paper, or Scissors correctly!")
else:
    print("Computer wins!")
    print("You lose!")
    print("Better luck next time!")
print("----------------------------------")
```
