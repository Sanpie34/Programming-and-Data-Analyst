```python
def move_item(lst, from_pos, to_pos, one_based=True):
    """
    Move an element in `lst` from `from_pos` to `to_pos`.
    By default positions are 1-based (set one_based=False for 0-based).
    Operation is in-place and the list is also returned.
    """
    if one_based:
        from_idx = from_pos - 1
        to_idx = to_pos - 1
    else:
        from_idx, to_idx = from_pos, to_pos

    if from_idx == to_idx:
        return lst
    if not (0 <= from_idx < len(lst)):
        raise IndexError("from_pos out of range")
    if not (0 <= to_idx <= len(lst)):
        raise IndexError("to_pos out of range")

    item = lst.pop(from_idx)
    lst.insert(to_idx, item)
    return lst

# Example with 10 items, move item at position 5 to position 1 (1-based)
#sample = ['a','b','c','d','e','f','g','h','i','j']
#print("before:", sample)
#move_item(sample, 5, 1)  # moves 'e' to front
#print("after: ", sample)

# If you want to operate on your existing list (e.g. my_istri), call:
move_item(my_istri, 11, 2)
```
