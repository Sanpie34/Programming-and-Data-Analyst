### Change the order list/array of my wife priority

```python
def move_item(lst, from_pos, to_pos, one_based=True):
    """
    Move an element in `lst` from `from_pos` to `to_pos`.
    By default positions are 1-based (set one_based=False for 0-based).
    Operation is in-place and the list is also returned.
    """
    if one_based:
        from_idx = from_pos - 1
        to_idx = to_pos - 1
    else:
        from_idx, to_idx = from_pos, to_pos

    if from_idx == to_idx:
        return lst
    if not (0 <= from_idx < len(lst)):
        raise IndexError("from_pos out of range")
    if not (0 <= to_idx <= len(lst)):
        raise IndexError("to_pos out of range")

    item = lst.pop(from_idx)
    lst.insert(to_idx, item)
    return lst

# Example with 10 items, move item at position 5 to position 1 (1-based)
#sample = ['a','b','c','d','e','f','g','h','i','j']
#print("before:", sample)
#move_item(sample, 5, 1)  # moves 'e' to front
#print("after: ", sample)

# If you want to operate on your existing list (e.g. my_istri), call:
move_item(my_istri, 11, 2)
move_item(my_istri, 4, 3)
```

# Gacha Pull Simulator

### Gacha Pull Simulator with same probability

```python
def gacha(my_istri):
    chosen_istri = np.random.choice(my_istri)
    return chosen_istri

gacha_pulls = int(input("How many gacha pulls do you want to do? "))
for _ in range(gacha_pulls):
    result = gacha(my_istri)
    print(f"You pulled: {result}")




```

### Gacha Pull Simulator with different probability

```python
def limited_gacha(my_istri):
    probabilities = [0.05 if istri in ["Ye Shunguang", "Castorice", "Nagusa", "Firefly"] else 0.80/(len(my_istri)-4) for istri in my_istri]
    chosen_istri = np.random.choice(my_istri, p=probabilities)
    return chosen_istri

gacha_pulls = int(input("How many limited gacha pulls do you want to do? "))
for _ in range(gacha_pulls):
    result = limited_gacha(my_istri)
    print(f"You pulled: {result}")
```

### Gacha Pull Simulator of limited 5 star with 3 and 4 star chara

```python
three_star = ["trash" for _ in range(50)]
four_star = ["Ibuki", "Miyako", "Hoshino", "Ju Fufu", "Yi Xuan", "Mika", "Iroha", "Aglaea", "Tribbie", "Yunli", "Lingsha", "Acheron", "Kafka", "Miyu", "Aoi", "Seele", "Bronya", "Clara", "Huohuo", "Xueyi", "Ayaka", "Raiden Ei", "Nilou", "Kokomi", "Lumine", "Charlotte", "Aris", "Maki", "Hare", "Himari", "Kanao", "Kayoko", "Fubuki"]
gacha = my_istri + three_star + four_star
def gacha_limited_my_istri(gacha):
    probabilities = []
    for istri in gacha:
        if istri in ["Ye Shunguang", "Castorice", "Nagusa", "Firefly"]:
            probabilities.append(0.0125)
        elif istri in three_star:
            probabilities.append(0.75/(len(three_star)))
        elif istri in four_star:
            probabilities.append(0.175/(len(four_star)))
        else:
            probabilities.append(0.025/(len(gacha) - len(three_star) - len(four_star) - 4))
    chosen_istri = np.random.choice(gacha, p=probabilities)
    return chosen_istri

gacha_pulls = int(input("How many gacha pulls do you want to do? "))
for _ in range(gacha_pulls):
    result = gacha_limited_my_istri(gacha)
    if result in three_star:
        print(f"You pulled: {result}")
    elif result in four_star:
        print("")
        print("You pulled a 4 star character!")
        print(f"You pulled: {result}")
        print("")
    elif result in ["Ye Shunguang", "Castorice", "Nagusa", "Firefly"]:
        print("--------------------------------------------------")
        print("Congratulations! She want to be your wife!")
        print(f"You pulled: {result}")
        print("--------------------------------------------------")
    else:
        print("...........................................")
        print("Congrats! She want to be your girlfriend.")
        print(f"You pulled: {result}")
        print("...........................................")
```
